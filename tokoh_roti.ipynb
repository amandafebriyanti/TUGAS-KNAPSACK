```python
def knapsack_dp(weights, profits, capacity):
    n = len(weights)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        for w in range(1, capacity + 1):
            if weights[i-1] <= w:
                dp[i][w] = max(dp[i-1][w], dp[i-1][w - weights[i-1]] + profits[i-1])
            else:
                dp[i][w] = dp[i-1][w]

    selected = []
    w = capacity
    for i in range(n, 0, -1):
        if dp[i][w] != dp[i-1][w]:
            selected.append(i-1)
            w -= weights[i-1]

    selected.reverse()
    return dp[n][capacity], selected

# Data roti
jenis_roti = ["Roti Tawar", "Roti Gandum", "Roti Cokelat", "Roti Keju", "Roti Pisang",
              "Roti Sosis", "Roti Kismis", "Roti Pandan", "Roti Sobek", "Roti Manis"]
berat = [8, 7, 6, 6, 5, 7, 4, 5, 6, 4]
keuntungan = [120000, 110000, 105000, 115000, 95000, 130000, 85000, 90000, 100000, 80000]
kapasitas = 20

# Eksekusi
max_profit, selected_idx = knapsack_dp(berat, keuntungan, kapasitas)

# Output
print("Kapasitas Maksimum Oven:", kapasitas, "loyang")
print("Total Keuntungan Maksimum: Rp", max_profit)
print("Roti yang dipilih:")
total_berat = 0
for idx in selected_idx:
    print(f"  - {jenis_roti[idx]} (Berat: {berat[idx]}, Keuntungan: Rp{keuntungan[idx]})")
    total_berat += berat[idx]
print("Total Berat Terpilih:", total_berat, "loyang")
```
